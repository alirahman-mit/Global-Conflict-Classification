## 1. Apa itu Perulangan (Loop)?

Coba bayangkan kita harus mencetak tulisan "Baris 1" sampai "Baris 5" satu per satu. Kalau harus dilakukan manual, bayangkan kalau harus sampai "Baris 1000" — pasti sangat melelahkan dan gampang salah ketik!

**Loop (perulangan)** membuat komputer mengulang sebuah perintah secara otomatis, cepat, dan rapi tanpa salah ketik.

Alur kerja sebuah loop kurang lebih begini:

```
▶ Mulai  →  Cek Kondisi  →  Ulangi / Berhenti
```

Selama kondisinya terpenuhi, loop akan terus berjalan. Begitu kondisinya tidak terpenuhi lagi, loop berhenti.

Yuk bandingkan langsung caranya di bawah 👇

````python
# Tanpa loop -- capek, dan gampang salah ketik
print("Baris 1")
print("Baris 2")
print("Baris 3")
print("Baris 4")
print("Baris 5")
````

````python
# Dengan loop -- singkat, rapi, dan gampang diubah
for i in range(1, 6+1):
    print("Baris", i)

# coba ganti angka 6 menjadi 1001 untuk cetak 1000 baris,
# kodenya tetap cuma 2 baris!
````

## 2. `for` Loop & `range()`

`for` dipakai saat kita **sudah tahu** berapa kali harus mengulang, misalnya untuk setiap item dalam daftar, atau angka dari 0 sampai 5.

### Anatomi sebuah `for` loop

```python
for i in range(0, 5):
    print(i)
```

| Bagian | Penjelasan |
|---|---|
| `for` | kata kunci yang memulai perulangan |
| `i` | variabel yang nilainya berubah tiap putaran (disebut juga *counter*) |
| `in` | penghubung antara variabel dengan sumber data |
| `range(0, 5)` | sumber data / daftar nilai yang akan diulang |
| `:` | menandai awal blok kode yang akan diulang (badan loop) |

### `range(start, stop, step)`
- `stop` **tidak ikut** dihitung
- `step` menentukan lompatannya (boleh dilewatkan, defaultnya 1)
- `step` boleh negatif untuk menghitung mundur!

````python
# range(stop) -- mulai dari 0, lompat 1
for i in range(5):
    print(i)
````

````python
# range(start, stop) -- mulai dari angka tertentu
for i in range(2, 8):
    print(i)
````

````python
# range(start, stop, step) -- melompat 2 angka setiap kali
for i in range(0, 10, 2):
    print(i)
````

````python
for i in range(0, 10, 1):
    print(i)
````

````python
# range dengan step negatif -- menghitung mundur
for i in range(10, 0, -1):
    print(i)
````

### `for` juga bisa langsung ke list dan string, lho!

Tidak selalu harus lewat `range()` — `for` bisa langsung mengulang setiap item dalam list, atau setiap huruf dalam sebuah teks.

````python
buah = ["Apel", "Jeruk", "Mangga"]
for nama in buah:
    print(nama)
````

````python
for huruf in "Python":
    print(huruf)
````

## 3. `while` Loop

`while` dipakai saat jumlah pengulangan **belum pasti**, tergantung sebuah syarat yang bisa berubah.

### Anatomi sebuah `while` loop

```python
i = 0
while i < 5:
    print(i)
    i += 1
```

| Bagian | Penjelasan |
|---|---|
| `i = 0` | nilai awal, disiapkan sebelum loop dimulai |
| `while i < 5` | syarat yang dicek setiap kali sebelum satu putaran dijalankan |
| `print(i)` | badan loop, dijalankan selama syarat masih `True` |
| `i += 1` | update variabel — **WAJIB ada**, agar suatu saat syaratnya jadi `False` |

⚠️ Kalau kita lupa menulis baris update (`i += 1`), loop akan berjalan **selamanya** (disebut *infinite loop*)!

````python
i = 0
while i < 5:
    print(i)
    i += 0.5
````

````python
baterai = 100
while baterai > 0:
    print("Baterai:", baterai)
    baterai -= 20
````

## 4. `break` — Menghentikan Loop

`break` langsung menghentikan loop **sepenuhnya**, walaupun belum selesai — seperti tombol berhenti darurat 🛑.

````python
for angka in range(1, 10):
    if angka == 3:
        break
    print(angka)

# output: 1  2
# begitu angka == 3, loop langsung berhenti total
````

## 5. `continue` — Melompati Iterasi

`continue` hanya melompati langkah saat ini, lalu **lanjut** ke iterasi berikutnya — beda dengan `break` yang berhenti total.

````python
for angka in range(0, 10):
    if angka == 3:
        continue
    print(angka)

# output: 1  2  4
# angka 3 dilewati, tapi loop tetap lanjut sampai selesai
````

````python
for angka in range(0, 10):
    if angka%2 == 1: #skip angka ganjil
        continue
    print(angka)
````

### Ringkasan `break` vs `continue`

| | `break` | `continue` |
|---|---|---|
| Efek | Menghentikan loop sepenuhnya | Hanya melompati iterasi saat ini |
| Sisa iterasi | Tidak dijalankan lagi | Loop tetap lanjut ke putaran berikutnya |
| Ibarat | "Berhenti total, langsung keluar" | "Skip dulu, tapi tetap jalan terus" |
| Cocok untuk | Berhenti begitu sudah ketemu yang dicari | Melewati data yang tidak relevan |

## 6. Nested Loop (Loop Bersarang)

Loop luar berjalan lebih lambat, loop dalam berjalan penuh untuk **setiap** langkah loop luar. Cocok untuk membuat pola grid, tabel, atau papan catur.

````python
for baris in range(3):
    for kolom in range(3):
        print("*", end=" ")
    print()
````

### Contoh: Tabel Perkalian

Loop luar (`i`) menentukan baris, loop dalam (`j`) menentukan kolom — kombinasi keduanya membentuk grid.

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end="  ")
    print()
````

````python
import string

end = 10
huruf = string.ascii_lowercase

for baris in range(1, end):
  count = 0
  for kolom in huruf:
    count += 1
    if count == baris:
      break
    print(kolom, end='')
  print()
````

## 8. Pola Akumulator

Siapkan variabel `total` bernilai `0` sebelum loop dimulai, lalu tambahkan nilainya setiap putaran. Pola ini banyak dipakai untuk menghitung total belanja, skor game, atau jumlah nilai ujian.

````python
total = 0
for angka in range(1, 6):
    total += angka
print(total)  # 1+2+3+4+5 = 15
````

Pola akumulator juga tidak harus untuk menjumlahkan angka. Contoh lain: menghitung berapa banyak huruf vokal dalam sebuah kata.

````python
kata = "pemrograman"
jumlah_vokal = 0
vokal = "aiueo"

for huruf in kata:
    if huruf in vokal:
        jumlah_vokal += 1

print("Jumlah huruf vokal:", jumlah_vokal)
````

## 11. Studi Kasus: Loop di Dunia Nyata (IoT & AI)

### 🌡️ Sensor IoT
Alat pemantau suhu di dunia nyata terus membaca data setiap beberapa detik menggunakan `while True`. Supaya notebook ini tidak berjalan selamanya, kita simulasikan dengan **batas jumlah pembacaan** memakai `while` + `break`.

````python
import random

jumlah_bacaan = 0
batas_bacaan = 5  # simulasi: sensor kita batasi baca 5 kali saja

while True:
    suhu = round(random.uniform(24.0, 32.0), 1)
    print(f"Bacaan ke-{jumlah_bacaan + 1} -> Suhu: {suhu}°C")
    jumlah_bacaan += 1
    if jumlah_bacaan >= batas_bacaan:
        print("Simulasi selesai, sensor berhenti membaca.")
        break
````

### 🤖 Melatih Model AI
Model AI belajar dengan mengulang proses training berkali-kali — tiap putaran disebut satu **epoch**. Di bawah ini kita simulasikan *loss* (tingkat kesalahan model) yang semakin mengecil setiap epoch.

````python
jumlah_epoch = 10
loss = 1.0  # loss awal (semakin kecil semakin bagus)

for epoch in range(1, jumlah_epoch + 1):
    loss = loss * 0.8  # setiap epoch, loss berkurang 20%
    print(f"Epoch {epoch}/{jumlah_epoch} -> loss: {loss:.4f}")

print("Training selesai!")
````
